```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np
```

```python
x_data = [1.0, 1.8, 3.0, 4.1, 5.2, 6.0]
y_data = [1, 1.3, 2.2, 2.5, 2.8, 3.6]
```

```python
x = np.array(x_data)
y = np.array(y_data)

print(f"x = {x}")
print(f"y = {y}")
```

```python
plt.figure(figsize=(10, 4))

plt.scatter(x_data, y_data, color='red')
plt.xlabel("Data x")
plt.ylabel("Data y")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
x = x.reshape(-1, 1)
y = y.reshape(-1, 1)

print(f"x = {x}")
print(f"y = {y}")
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
model.predict([[5.5]])
```

```python
prd = model.predict(x)

plt.figure(figsize=(10, 4))

plt.scatter(x_data, y_data, color='red')
plt.plot(x, prd, color='blue')
plt.scatter(x, prd, color='orange', marker='s')

plt.xlabel("Data x")
plt.ylabel("Data y")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
source = 'datasets/fruit_machine.csv'

fruit_df = pd.read_csv(source)
fruit_df
```

```python
size = fruit_df['size']
fruit_name = fruit_df['fruit_name']
```

```python
fruit_df.loc[fruit_df["fruit_name"] == 'lemon', 'class'] = '0'
fruit_df.loc[fruit_df["fruit_name"] == 'orange', 'class'] = '1'
fruit_df
```

```python
y = fruit_df['class']
y
```

```python
x = np.array(size).reshape(-1, 1)
y = np.array(y).reshape(-1, 1)

print(x)
print(y)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(x, y)
```

```python
new_size = 25
y_predict = model.predict([[new_size]])
print(f"Result : ", y_predict[0])
```

```python
class_names = ['lemon', 'orange']
```

```python
new_size = 25
y_predict = model.predict([[new_size]])

prd = class_names[int(y_predict[0])]
print(f"Result : ", prd)
```

```python
while 1:
  size = int(input("Enter size : "))
  if size == 0:
    print("Done!")
    break
  y_predict = model.predict([[size]])
  result = class_names[y_predict[0]]

  print(f"Result : ", result)

  if (result == 'lemon'):
    print('<--- push to lemon basket')
  else:
    print('---> push to orange basket')
```
